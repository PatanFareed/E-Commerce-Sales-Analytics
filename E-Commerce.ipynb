step 3B Feature Engineering
--

```python
df["year"] = df["Order_Date"].dt.year
df["Month_Name"] = df["Order_Date"].dt.month_name()
df["Month_Num"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Quater"] = df["Order_Date"].dt.quarter
df["Week"] = df["Order_Date"].dt.weekday
```

```python
df["Profit_Margin_%"] = df["Profit"]/df["Net_Sales"]*100
```

```python
df["Quantity"].sum()
```

```python
sales_metrics = df[["Order_ID","Customer_ID","Quantity","Cost","Unit_Price","Discount_%"]]
sales_metrics
```

```python
sales_metrics["Gross_Sales"] = sales_metrics["Quantity"]*sales_metrics["Unit_Price"]
sales_metrics["Discount_Amount"] = sales_metrics["Gross_Sales"]*sales_metrics["Discount_%"]/100
sales_metrics["Net_Sales"] = sales_metrics["Gross_Sales"]-sales_metrics["Discount_Amount"]
sales_metrics["Profit"] = sales_metrics["Net_Sales"]-sales_metrics["Cost"]
sales_metrics["Profit_Margin"] = sales_metrics["Profit"]/sales_metrics["Net_Sales"]*100
```

```python
sales_metrics
```

```python
cust_metrics = df.groupby(["Customer_ID","Customer_Name"]).agg(
        Total_rders = ("Order_ID","count"),
        Total_Sales = ("Net_Sales","sum"),
        Avg_ord_val = ("Net_Sales","mean")
)
cust_metrics
```
